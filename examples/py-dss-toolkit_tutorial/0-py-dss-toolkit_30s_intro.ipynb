```python
import os
# Colab: clone goes to /content. Local: use current dir.
base = "/content" if os.path.exists("/content") else os.getcwd()
dss_file_path = os.path.join(base, "opendss-python-examples", "feeder_models", "EPRITestCircuits", "ckt5", "Master_ckt5.dss")
print(f"Using: {dss_file_path}")
```

## 2. Load the model and run a power flow

```python
import py_dss_interface
from py_dss_toolkit import dss_tools

dss = py_dss_interface.DSS()
dss_tools.update_dss(dss)

dss.text(f"compile [{dss_file_path}]")
dss_tools.simulation.solve_snapshot(max_control_iter=20)

print(f"Circuit: {dss.circuit.name}")
print(f"Converged: {dss.solution.converged}")
```

## 3. Explore the model — segments

```python
segments_df = dss_tools.model.segments_df
print(f"Segments: {len(segments_df)}")
segments_df.head(10)
```

## 4. Voltages in DataFrames

```python
vmag_df, vang_df = dss_tools.results.voltage_ln_nodes
print("Voltage magnitude (V) — first 5 buses:")
vmag_df.head()
```

## 5. Voltage profile

Plot voltage magnitude along the feeder path (distance from source).

```python
dss_tools.interactive_view.voltage_profile()
```

## 6. Plot active power on the circuit map

```python
dss_tools.interactive_view.circuit_plot(parameter="active power", title="Active Power [kW]")
```
